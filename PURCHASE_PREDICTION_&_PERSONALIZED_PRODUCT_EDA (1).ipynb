```python
# ============================================================
# CUSTOMER PURCHASE PREDICTION AND PERSONALIZED PRODUCT
# RECOMMENDATION USING MACHINE LEARNING
#
# EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)

print("Libraries imported successfully!")


# ============================================================
# 2. CONNECT GOOGLE DRIVE
# ============================================================

from google.colab import drive

drive.mount('/content/drive')


# ============================================================
# 3. DATASET PATH
# ============================================================

BASE_PATH = "/content/drive/MyDrive/Colab/archive"

events_path = f"{BASE_PATH}/events.csv"
category_path = f"{BASE_PATH}/category_tree.csv"

print("Dataset path:")
print(events_path)


# ============================================================
# 4. LOAD EVENTS DATASET
# ============================================================

events = pd.read_csv(events_path)

print("\nDataset loaded successfully!")
print("Dataset Shape:", events.shape)


# ============================================================
# 5. DISPLAY FIRST FIVE ROWS
# ============================================================

print("\nFirst five rows:")
display(events.head())


# ============================================================
# 6. DATASET INFORMATION
# ============================================================

print("\nDataset Information:")
events.info()


# ============================================================
# 7. BASIC DATASET DESCRIPTION
# ============================================================

print("\nNumber of Rows:", events.shape[0])
print("Number of Columns:", events.shape[1])

print("\nColumn Names:")
print(events.columns.tolist())

print("\nData Types:")
print(events.dtypes)


# ============================================================
# 8. DATASET SUMMARY TABLE
# ============================================================

summary = pd.DataFrame({
    "EDA Item": [
        "Rows",
        "Columns",
        "Unique Customers",
        "Unique Products",
        "Event Types",
        "Missing Values",
        "Duplicate Rows"
    ],

    "Result": [
        f"{events.shape[0]:,}",
        events.shape[1],
        f"{events['visitorid'].nunique():,}",
        f"{events['itemid'].nunique():,}",
        events['event'].nunique(),
        f"{events.isnull().sum().sum():,}",
        f"{events.duplicated().sum():,}"
    ]
})

print("\n========== DATASET SUMMARY ==========")
display(summary)


# ============================================================
# 9. MISSING VALUE ANALYSIS
# ============================================================

missing_values = events.isnull().sum()

missing_table = pd.DataFrame({
    "Column": missing_values.index,
    "Missing Values": missing_values.values,
    "Percentage": (
        missing_values.values / len(events) * 100
    ).round(2)
})

print("\n========== MISSING VALUE ANALYSIS ==========")
display(missing_table)


# ============================================================
# 10. DUPLICATE ANALYSIS
# ============================================================

duplicate_count = events.duplicated().sum()

print("\n========== DUPLICATE ANALYSIS ==========")
print("Number of duplicate rows:", duplicate_count)


# ============================================================
# 11. CONVERT TIMESTAMP
# ============================================================

events['datetime'] = pd.to_datetime(
    events['timestamp'],
    unit='ms'
)

print("\nTimestamp converted successfully!")

print("\nTime Range:")
print("Start:", events['datetime'].min())
print("End:", events['datetime'].max())


# ============================================================
# 12. CUSTOMER EVENT DISTRIBUTION
# ============================================================

event_counts = events['event'].value_counts()

print("\n========== CUSTOMER EVENT DISTRIBUTION ==========")
display(event_counts)


# ============================================================
# 13. EVENT DISTRIBUTION GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

event_counts.plot(kind='bar')

plt.title("Customer Event Distribution")
plt.xlabel("Event Type")
plt.ylabel("Number of Events")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 14. EVENT PERCENTAGES
# ============================================================

event_percentages = (
    events['event']
    .value_counts(normalize=True) * 100
)

print("\n========== EVENT PERCENTAGES ==========")
display(event_percentages.round(2))


# ============================================================
# 15. CUSTOMER ACTIVITY ANALYSIS
# ============================================================

customer_activity = events.groupby('visitorid').size()

print("\n========== CUSTOMER ACTIVITY STATISTICS ==========")
display(customer_activity.describe())


# ============================================================
# 16. CUSTOMER ACTIVITY DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

plt.hist(
    customer_activity[customer_activity <= 100],
    bins=50
)

plt.title("Distribution of Customer Activity")
plt.xlabel("Number of Interactions per Customer")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()


# ============================================================
# 17. TOP 10 MOST VIEWED PRODUCTS
# ============================================================

views = events[
    events['event'] == 'view'
]

top_viewed = (
    views['itemid']
    .value_counts()
    .head(10)
)

print("\n========== TOP 10 MOST VIEWED PRODUCTS ==========")
display(top_viewed)


# ============================================================
# 18. MOST VIEWED PRODUCTS GRAPH
# ============================================================

plt.figure(figsize=(10, 5))

top_viewed.sort_values().plot(kind='barh')

plt.title("Top 10 Most Viewed Products")
plt.xlabel("Number of Views")
plt.ylabel("Product ID")

plt.tight_layout()
plt.show()


# ============================================================
# 19. TOP 10 MOST PURCHASED PRODUCTS
# ============================================================

transactions = events[
    events['event'] == 'transaction'
]

top_purchased = (
    transactions['itemid']
    .value_counts()
    .head(10)
)

print("\n========== TOP 10 MOST PURCHASED PRODUCTS ==========")
display(top_purchased)


# ============================================================
# 20. MOST PURCHASED PRODUCTS GRAPH
# ============================================================

plt.figure(figsize=(10, 5))

top_purchased.sort_values().plot(kind='barh')

plt.title("Top 10 Most Purchased Products")
plt.xlabel("Number of Purchases")
plt.ylabel("Product ID")

plt.tight_layout()
plt.show()


# ============================================================
# 21. TOP 10 PRODUCTS ADDED TO CART
# ============================================================

cart_events = events[
    events['event'] == 'addtocart'
]

top_cart = (
    cart_events['itemid']
    .value_counts()
    .head(10)
)

print("\n========== TOP 10 PRODUCTS ADDED TO CART ==========")
display(top_cart)


# ============================================================
# 22. ADD-TO-CART PRODUCTS GRAPH
# ============================================================

plt.figure(figsize=(10, 5))

top_cart.sort_values().plot(kind='barh')

plt.title("Top 10 Products Added to Cart")
plt.xlabel("Number of Add-to-Cart Events")
plt.ylabel("Product ID")

plt.tight_layout()
plt.show()


# ============================================================
# 23. CREATE DATE AND TIME FEATURES
# ============================================================

events['date'] = events['datetime'].dt.date
events['hour'] = events['datetime'].dt.hour
events['day_of_week'] = events['datetime'].dt.day_name()

print("\nDate and time features created successfully!")


# ============================================================
# 24. EVENTS OVER TIME
# ============================================================

daily_events = (
    events.groupby('date')
    .size()
)

plt.figure(figsize=(14, 5))

daily_events.plot()

plt.title("Customer Events Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Events")

plt.tight_layout()
plt.show()


# ============================================================
# 25. EVENT TYPES OVER TIME
# ============================================================

daily_event_type = (
    events.groupby(['date', 'event'])
    .size()
    .unstack(fill_value=0)
)

plt.figure(figsize=(14, 6))

for event_type in daily_event_type.columns:
    plt.plot(
        daily_event_type.index,
        daily_event_type[event_type],
        label=event_type
    )

plt.title("Customer Event Types Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Events")
plt.legend()

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# 26. CUSTOMER ACTIVITY BY HOUR
# ============================================================

hourly_events = (
    events['hour']
    .value_counts()
    .sort_index()
)

print("\n========== HOURLY CUSTOMER ACTIVITY ==========")
display(hourly_events)


# ============================================================
# 27. HOURLY ACTIVITY GRAPH
# ============================================================

plt.figure(figsize=(10, 5))

hourly_events.plot(kind='bar')

plt.title("Customer Activity by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Events")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 28. CUSTOMER PURCHASE STATUS
# ============================================================

total_customers = (
    events['visitorid']
    .nunique()
)

customers_with_purchase = (
    events.loc[
        events['event'] == 'transaction',
        'visitorid'
    ].nunique()
)

customers_without_purchase = (
    total_customers -
    customers_with_purchase
)

print("\n========== CUSTOMER PURCHASE STATUS ==========")

print("Total Customers:",
      f"{total_customers:,}")

print("Customers Who Purchased:",
      f"{customers_with_purchase:,}")

print("Customers Without Purchase:",
      f"{customers_without_purchase:,}")


# ============================================================
# 29. PURCHASE STATUS GRAPH
# ============================================================

purchase_status = pd.Series({
    'Purchased': customers_with_purchase,
    'No Purchase': customers_without_purchase
})

plt.figure(figsize=(7, 5))

purchase_status.plot(kind='bar')

plt.title("Customers With and Without Purchases")
plt.xlabel("Customer Status")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 30. CUSTOMER-LEVEL FEATURE ENGINEERING
# ============================================================

customer_features = events.groupby('visitorid').agg(

    total_interactions=('event', 'count'),

    total_views=(
        'event',
        lambda x: (x == 'view').sum()
    ),

    total_cart_adds=(
        'event',
        lambda x: (x == 'addtocart').sum()
    ),

    total_purchases=(
        'event',
        lambda x: (x == 'transaction').sum()
    ),

    unique_products=(
        'itemid',
        'nunique'
    )
)

print("\n========== CUSTOMER FEATURES ==========")
display(customer_features.head())


# ============================================================
# 31. PURCHASE RATE
# ============================================================

customer_features['purchase_rate'] = (
    customer_features['total_purchases'] /
    customer_features['total_interactions']
)

print("\nCustomer features with purchase rate:")
display(customer_features.head())


# ============================================================
# 32. PURCHASE RATE DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

plt.hist(
    customer_features['purchase_rate'],
    bins=50
)

plt.title("Distribution of Customer Purchase Rate")
plt.xlabel("Purchase Rate")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()


# ============================================================
# 33. VIEWS VS PURCHASES
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    customer_features['total_views'],
    customer_features['total_purchases'],
    alpha=0.3
)

plt.title("Views vs Purchases")
plt.xlabel("Total Views")
plt.ylabel("Total Purchases")

plt.tight_layout()
plt.show()


# ============================================================
# 34. ADD-TO-CART VS PURCHASES
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    customer_features['total_cart_adds'],
    customer_features['total_purchases'],
    alpha=0.3
)

plt.title("Add-to-Cart Actions vs Purchases")
plt.xlabel("Total Add-to-Cart Actions")
plt.ylabel("Total Purchases")

plt.tight_layout()
plt.show()


# ============================================================
# 35. CORRELATION ANALYSIS
# ============================================================

correlation_features = customer_features[
    [
        'total_interactions',
        'total_views',
        'total_cart_adds',
        'total_purchases',
        'unique_products',
        'purchase_rate'
    ]
]

correlation_matrix = (
    correlation_features.corr()
)

print("\n========== CORRELATION MATRIX ==========")
display(correlation_matrix)


# ============================================================
# 36. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title(
    "Correlation Heatmap of Customer Behavioral Features"
)

plt.tight_layout()
plt.show()


# ============================================================
# 37. LOAD CATEGORY TREE
# ============================================================

category_tree = pd.read_csv(
    category_path
)

print("\n========== CATEGORY DATASET ==========")

print("Shape:",
      category_tree.shape)

print("\nColumns:")
print(category_tree.columns.tolist())

display(category_tree.head())


# ============================================================
# 38. CATEGORY DATA QUALITY
# ============================================================

print("\nNumber of Categories:",
      category_tree['categoryid'].nunique())

print("\nMissing Values:")
display(category_tree.isnull().sum())

print("\nDuplicate Rows:",
      category_tree.duplicated().sum())


# ============================================================
# 39. FINAL EDA FINDINGS
# ============================================================

print("\n")
print("=" * 60)
print("FINAL EDA FINDINGS")
print("=" * 60)

print(
    f"\n1. The dataset contains "
    f"{len(events):,} customer interaction records."
)

print(
    f"2. There are "
    f"{events['visitorid'].nunique():,} unique customers."
)

print(
    f"3. There are "
    f"{events['itemid'].nunique():,} unique products."
)

print(
    f"4. The dataset contains "
    f"{events['event'].nunique()} types of customer events."
)

print("\n5. Event Distribution:")
print(events['event'].value_counts())

print(
    f"\n6. Number of duplicate rows: "
    f"{events.duplicated().sum():,}"
)

print("\n7. Missing Values:")
print(events.isnull().sum())

print(
    f"\n8. Customers who made at least one purchase: "
    f"{customers_with_purchase:,}"
)

print(
    f"9. Customers without a purchase: "
    f"{customers_without_purchase:,}"
)

print(
    f"\n10. Time range: "
    f"{events['datetime'].min()} "
    f"to "
    f"{events['datetime'].max()}"
)


# ============================================================
# 40. IMPORTANT DATA QUALITY OBSERVATION
# ============================================================

print("\n")
print("=" * 60)
print("IMPORTANT DATA-QUALITY OBSERVATION")
print("=" * 60)

print("""
The transactionid column contains many missing values.
This is expected because not every customer interaction
results in a completed purchase.

View and add-to-cart events can occur without a transaction ID.
Therefore, missing transaction IDs should not automatically
be treated as invalid records or deleted during EDA.
""")

print("\nEDA COMPLETED SUCCESSFULLY!")
```
